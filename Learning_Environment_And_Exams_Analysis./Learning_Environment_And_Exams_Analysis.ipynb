```python
import os
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# Xác định đường dẫn tương đối tới các thư mục
DATA_PATH = os.path.join("..", "data", "Exam_Score_Prediction_Cleaned.csv")
RESULT_PATH = os.path.join("..", "result", "PhanTich_Moitruong.png")

# Cấu hình giao diện đồ thị
plt.style.use("seaborn-v0_8-whitegrid")
plt.rcParams["font.sans-serif"] = "DejaVu Sans"
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
fig.suptitle(
    "PHÂN TÍCH MÔI TRƯỜNG HỌC TẬP & KỲ THI TỚI ĐIỂM SỐ (EXAM SCORE)",
    fontsize=16,
    fontweight="bold",
)

# Đọc dữ liệu từ thư mục data
df = pd.read_csv(DATA_PATH)

# ---------------------------------------------------------
# 1. FACILITY RATING vs EXAM SCORE
# ---------------------------------------------------------
facility_avg = (
    df.groupby("Facility_Rating")["Exam_Score"]
    .mean()
    .reset_index()
    .sort_values(by="Facility_Rating")
)
sns.barplot(
    ax=axes[0, 0],
    data=facility_avg,
    x="Facility_Rating",
    y="Exam_Score",
    palette="Blues_d",
)
axes[0, 0].set_title(
    "1. Điểm trung bình theo Mức đánh giá Cơ sở vật chất",
    fontsize=12,
    fontweight="bold",
)
axes[0, 0].set_xlabel("Facility Rating (Đánh giá CSVC)")
axes[0, 0].set_ylabel("Điểm thi trung bình (Exam Score)")
for p in axes[0, 0].patches:
    axes[0, 0].annotate(
        f"{p.get_height():.2f}",
        (p.get_x() + p.get_width() / 2.0, p.get_height()),
        ha="center",
        va="center",
        xytext=(0, 5),
        textcoords="offset points",
    )

sns.countplot(
    ax=axes[0, 1],
    data=df,
    x="Facility_Rating",
    palette="Blues_d",
    order=sorted(df["Facility_Rating"].unique()),
)
axes[0, 1].set_title(
    "2. Số lượng học sinh theo Mức đánh giá Cơ sở vật chất",
    fontsize=12,
    fontweight="bold",
)
axes[0, 1].set_xlabel("Facility Rating (Đánh giá CSVC)")
axes[0, 1].set_ylabel("Số lượng học sinh")
for p in axes[0, 1].patches:
    axes[0, 1].annotate(
        f"{int(p.get_height())}",
        (p.get_x() + p.get_width() / 2.0, p.get_height()),
        ha="center",
        va="center",
        xytext=(0, 5),
        textcoords="offset points",
    )

# ---------------------------------------------------------
# 2. EXAM DIFFICULTY vs EXAM SCORE
# ---------------------------------------------------------
diff_order = [
    cat
    for cat in ["Low", "Easy", "Medium", "High", "Hard"]
    if cat in df["Exam_Difficulty"].unique()
]
if not diff_order:
    diff_order = df["Exam_Difficulty"].unique()

diff_avg = (
    df.groupby("Exam_Difficulty")["Exam_Score"].mean().loc[diff_order].reset_index()
)
sns.barplot(
    ax=axes[1, 0],
    data=diff_avg,
    x="Exam_Difficulty",
    y="Exam_Score",
    palette="Oranges_d",
)
axes[1, 0].set_title(
    "3. Điểm trung bình theo Độ khó đề thi (Đề dễ / TB / Khó)",
    fontsize=12,
    fontweight="bold",
)
axes[1, 0].set_xlabel("Exam Difficulty (Độ khó đề thi)")
axes[1, 0].set_ylabel("Điểm thi trung bình (Exam Score)")
for p in axes[1, 0].patches:
    axes[1, 0].annotate(
        f"{p.get_height():.2f}",
        (p.get_x() + p.get_width() / 2.0, p.get_height()),
        ha="center",
        va="center",
        xytext=(0, 5),
        textcoords="offset points",
    )

sns.boxplot(
    ax=axes[1, 1],
    data=df,
    x="Exam_Difficulty",
    y="Exam_Score",
    order=diff_order,
    palette="Oranges_d",
)
axes[1, 1].set_title(
    "4. Phân bố điểm số (Boxplot) theo Độ khó đề thi",
    fontsize=12,
    fontweight="bold",
)
axes[1, 1].set_xlabel("Exam Difficulty (Độ khó đề thi)")
axes[1, 1].set_ylabel("Điểm thi (Exam Score)")

plt.tight_layout(rect=[0, 0, 1, 0.96])

# Lưu biểu đồ vào thư mục result
plt.savefig(RESULT_PATH, dpi=300)
plt.show()

# =========================================================
# OUTPUT: BẢNG SO SÁNH THỐNG KÊ
# =========================================================
print("=" * 80)
print("BẢNG SO SÁNH THỐNG KÊ CHI TIẾT DỮ LIỆU")
print("=" * 80)

bang_facility = df.groupby("Facility_Rating")["Exam_Score"].agg(
    So_Luong="count",
    Diem_Trung_Binh="mean",
    Diem_Trung_Vi="median",
    Do_Lech_Chuan="std",
)
print("\n--- 1. Bảng so sánh theo Facility Rating ---")
print(bang_facility.to_string())

bang_difficulty = df.groupby("Exam_Difficulty")["Exam_Score"].agg(
    So_Luong="count",
    Diem_Trung_Binh="mean",
    Diem_Trung_Vi="median",
    Do_Lech_Chuan="std",
)
print("\n--- 2. Bảng so sánh theo Exam Difficulty (Đề dễ / TB / Khó) ---")
print(bang_difficulty.to_string())
```
