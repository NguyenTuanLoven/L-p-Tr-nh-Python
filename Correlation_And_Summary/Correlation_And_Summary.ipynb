# NGƯỜI 5 — PHÂN TÍCH TƯƠNG QUAN & TỔNG HỢP
**Nội dung:** Tìm yếu tố có mối quan hệ mạnh nhất với `exam_score` và tổng hợp kết quả toàn nhóm.

```python
# 1. Đọc dữ liệu và tạo ma trận tương quan
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('Exam_Score_Prediction.csv')
corr = df.select_dtypes(include='number').corr()
print(corr)
```

```python
# 2. Vẽ biểu đồ Heatmap ma trận tương quan và lưu lại thành ảnh
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title("Ma trận tương quan giữa các biến")
plt.savefig('fig1_correlation_heatmap.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# 3. Sắp xếp mức độ tương quan với Exam Score
score_corr = corr['exam_score'].sort_values(ascending=False)
print(score_corr)
```

```python
# 4. Vẽ biểu đồ kiểm chứng các cặp biến chính và lưu lại thành ảnh
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
sns.regplot(data=df, x='study_hours', y='exam_score', ax=axes[0], scatter_kws={'alpha':0.3}, line_kws={'color':'red'})
axes[0].set_title('Study Hours vs Exam Score')

sns.regplot(data=df, x='class_attendance', y='exam_score', ax=axes[1], scatter_kws={'alpha':0.3}, line_kws={'color':'green'})
axes[1].set_title('Class Attendance vs Exam Score')

sns.regplot(data=df, x='sleep_hours', y='exam_score', ax=axes[2], scatter_kws={'alpha':0.3}, line_kws={'color':'blue'})
axes[2].set_title('Sleep Hours vs Exam Score')

plt.tight_layout()
plt.savefig('fig2_verification_plots.png', dpi=300, bbox_inches='tight')
plt.show()
```
